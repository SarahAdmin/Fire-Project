```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
fire = pd.read_csv('Fire_casualities_data.csv')
```

```python
fire.info()
```

```python
fire[' total_victims'] = fire[' total_victims'].replace({',':''},regex=True).astype(int)
```

```python
fire.head(10)
```

```python
sns.lineplot(data=fire,x='country',y=' total_victims',marker='o')
plt.title('Fire Casualities in the United Kingdom')
plt.xlabel('Region')
plt.ylabel('Total')
plt.show()
```

England is a most popular region in the United Kingdom with number of casualities. Wales has the least number of fire casualities by decreasing the total of casualities.

```python
sns.barplot(data=fire,x='country',y=' total_victims',hue='victim_type',errorbar=None)
plt.title('Fire Casualities in the United Kingdom')
plt.xlabel('Region')
plt.ylabel('Total')
plt.show()
```

England has more victim types on the bar chart than other regions.
Great Britian has only one type of total fire-related fatalities.
Scotland has second biggest number of casualities.
Wales which is a least popular region, has few number of casualities.

```python
sns.barplot(data=fire,x=' total_victims',y='victim_type',orient='h',palette='Set1',errcolor='rebeccapurple')
plt.xlabel('Total Victims')
plt.ylabel('Victim Type')
plt.show()
```

"Total casualities requiring hospital treatment" is a first biggest victim type. Second biggest victim type is "Total non-fatal casualities" shown on the horzontial bar chart. Other victim types has decreasing the number of total casualties to "Hospital severe".
